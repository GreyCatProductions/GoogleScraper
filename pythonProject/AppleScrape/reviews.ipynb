```python
from tkinter.ttk import Treeview

from tqdm import tqdm
from app_store_scraper import AppStore
import pickle
import pandas as pd
import os

initial_date = "2025-04-02"

all_links_apple = pd.read_csv(f"./data/Apple/{initial_date}/Apple.csv")

all_links_apple = all_links_apple[all_links_apple["review_count"] > 0] # TODO: Test!
all_links_apple = list(all_links_apple["link"])

reviews = []

os.makedirs("./data/Apple_Reviews/", exist_ok=True)
out_path = f"./data/Apple_Reviews/{initial_date}/"
os.makedirs(out_path, exist_ok=True)
```

```python
for i, app in tqdm(enumerate(all_links_apple), total=len(all_links_apple)):
    app_id = app.split("/")[-1][2:]
    temp = AppStore(country="de", app_name=app.split("/")[-2], app_id=app_id, log_level="DEBUG")
    temp.review(how_many=9999999, sleep=5)
    rev_list = temp.reviews
    if len(rev_list) > 0:
        rev = pd.DataFrame(temp.reviews) # TODO: Test!
        rev["id"] = app_id
        reviews.append(rev)
    if i % 1000 == 0:
        with open(out_path + "apple_reviews_" + str(i) + ".pkl", "wb") as f:
            pickle.dump(reviews, f)
        
            
with open(out_path + "apple_reviews_all.pkl", "wb") as f:
    pickle.dump(reviews, f) 
```
